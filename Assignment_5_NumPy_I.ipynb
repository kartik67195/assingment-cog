```python
# Q2. Reverse array and find most frequent value(s) and their indices
```

```python
# (a) Reverse
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed:", arr[::-1])

# (b) Most frequent value and all indices where it occurs
def mode_and_indices(a):
    values, counts = np.unique(a, return_counts=True)
    mode = values[counts == counts.max()]
    indices = {int(v): np.where(a == v)[0].tolist() for v in mode}
    return mode, indices

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1,1,1,2,3,4,2,4,3,3])

print("x modes and indices:", mode_and_indices(x))
print("y modes and indices:", mode_and_indices(y))
```

```python
# Q4. 25 evenly spaced numbers from 10 to 100
```

```python
karthik = np.linspace(10, 100, 25)
print(karthik)
print("Dimensions:", karthik.ndim)
print("Shape:", karthik.shape)
print("Total elements:", karthik.size)
print("Data type:", karthik.dtype)
print("Bytes:", karthik.nbytes)

# Reshape to a column (transpose of a row-shaped array)
row = karthik.reshape(1, 25)
column = row.reshape(25, 1)
print("Column shape:", column.shape)
print("Using .T:", row.T.shape)
# .T works for 2-D arrays; a 1-D array's .T remains 1-D.
```

```python
# Q5. 3 × 4 array: statistics, reshape and resize
```

```python
ucs420_karthik = np.array([[10, 20, 30, 40],
                             [50, 60, 70, 80],
                             [90, 15, 20, 35]])
print("Original array:\\n", ucs420_karthik)
print("Mean:", np.mean(ucs420_karthik))
print("Median:", np.median(ucs420_karthik))
print("Maximum:", np.max(ucs420_karthik))
print("Minimum:", np.min(ucs420_karthik))
print("Unique:", np.unique(ucs420_karthik))

reshaped_ucs420_karthik = ucs420_karthik.reshape(4, 3)
print("Reshaped 4x3:\\n", reshaped_ucs420_karthik)

resized_ucs420_karthik = np.resize(ucs420_karthik, (2, 3))
print("Resized 2x3:\\n", resized_ucs420_karthik)
```

```python
# Notes
# - reshape() changes shape without changing the number of elements.
# - np.resize() can repeat values when the requested shape needs more elements; here it truncates the array to six values.
# - Replace karthik and variable-name portions with your preferred name if needed.
```
