```python
# Q1. Vectorized sensor temperature processing
```

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected = temperature + 2
fahrenheit = (corrected * 9/5) + 32
above_32 = corrected[corrected > 32]
count_above_32 = np.sum(corrected > 32)

print("Corrected readings:", corrected)
print("Fahrenheit:", fahrenheit)
print("Corrected readings >32°C:", above_32)
print("Count >32°C:", count_above_32)
```

```python
# Explanation: NumPy vectorization applies operations to an array in compiled routines, avoiding Python-level loops. Boolean indexing filters values based on a condition; np.sum(condition) counts true entries.
```

```python
# Q2. Steps matrix statistics
```

```python
steps = np.array([[5000, 6200, 7100],
                   [8000, 7500, 9000],
                   [4500, 5100, 4800],
                   [9000, 8500, 9500]])

print("Total:", steps.sum())
print("Mean:", steps.mean())
print("Maximum:", steps.max())
print("Minimum:", steps.min())
print("Total for each day (axis=0):", steps.sum(axis=0))
print("Total for each user (axis=1):", steps.sum(axis=1))
max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum is at (user index, day index):", max_position)
print("Maximum value:", steps[max_position])
```

```python
# Q3. Array slicing, views, copies, reshape and flatten/ravel
```

```python
original = np.array([1, 2, 3, 4, 5, 6])
subset = original[1:4]
subset[0] = 999
print("After modifying slice:")
print("original:", original)
print("subset:", subset) # slice normally shares memory with original

copy_subset = original[1:4].copy()
copy_subset[0] = 500
print("After modifying copied slice:")
print("original:", original)
print("copy_subset:", copy_subset)
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
print("Matrix:\\n", matrix)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3 (zero-based indices):", matrix[1:3, 2:4])

flat_copy = matrix.flatten()
ravel_view = matrix.ravel()
print("flatten:", flat_copy)
print("ravel:", ravel_view)

ravel_view[0] = 1000
print("After modifying ravel:", matrix)

flat_copy[1] = 2000
print("After modifying flatten:", matrix) # matrix unaffected by flatten modification

print("shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
```

```python
# Q4. OLS coefficients for the cognitive-assistive score
```

```python
# The question's formula omits an intercept. The code below implements that exact formula using the three supplied feature columns. Since the matrix is not guaranteed to be invertible, pinv is a robust alternative; here we also attempt the inverse formula.
```

```python
X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])
y = np.array([40, 65, 30, 85])

print("Shape:", X.shape, "Dimensions:", X.ndim)
print("Transpose:\\n", X.T)
XtX = X.T @ X
print("X.T @ X:\\n", XtX)
print("Inverse (if available):")
try:
    inv_XtX = np.linalg.inv(XtX)
    beta = inv_XtX @ X.T @ y
except np.linalg.LinAlgError:
    print("X.T @ X is singular; using pseudoinverse.")
    beta = np.linalg.pinv(X) @ y

print("Coefficients [sleep, activity, stress]:", beta)
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```

```python
# Coefficient interpretation: In this linear model, each coefficient estimates the change in the predicted assistance score for a one-unit increase in that feature, while holding the other two features constant. This is an illustrative fit to only four observations, not a validated predictive system.
```
